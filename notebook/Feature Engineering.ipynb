```python
# Retrieve the stored transformed features and target variables
%store -r transformed_train_features
%store -r transformed_test_features
%store -r transformed_valid_features
%store -r train_labels
%store -r test_labels
%store -r valid_labels
```

```python
# Create a single categorical Wilderness_Area feature from the one-hot encoded columns

transformed_train_features['Wilderness_Area'] = transformed_train_features[
    ['Wilderness_Area_1', 'Wilderness_Area_2',
     'Wilderness_Area_3', 'Wilderness_Area_4']
].idxmax(axis=1).str.extract('(\\d+)').astype(int)

transformed_valid_features['Wilderness_Area'] = transformed_valid_features[
    ['Wilderness_Area_1', 'Wilderness_Area_2',
     'Wilderness_Area_3', 'Wilderness_Area_4']
].idxmax(axis=1).str.extract('(\\d+)').astype(int)

transformed_test_features['Wilderness_Area'] = transformed_test_features[
    ['Wilderness_Area_1', 'Wilderness_Area_2',
     'Wilderness_Area_3', 'Wilderness_Area_4']
].idxmax(axis=1).str.extract('(\\d+)').astype(int)
```

```python
# Create a ratio-based feature to represent the relationship between horizontal and vertical hydrology distances
transformed_train_features['Hydrology_Ratio'] = (
    transformed_train_features['Horizontal_Distance_To_Hydrology'] /
    (transformed_train_features['Vertical_Distance_To_Hydrology'].abs() + 1)
)

transformed_valid_features['Hydrology_Ratio'] = (
    transformed_valid_features['Horizontal_Distance_To_Hydrology'] /
    (transformed_valid_features['Vertical_Distance_To_Hydrology'].abs() + 1)
)

transformed_test_features['Hydrology_Ratio'] = (
    transformed_test_features['Horizontal_Distance_To_Hydrology'] /
    (transformed_test_features['Vertical_Distance_To_Hydrology'].abs() + 1)
)
```

```python
# Create a feature representing the difference between morning and afternoon hillshade
transformed_train_features['Hillshade_Diff'] = (
    transformed_train_features['Hillshade_9am'] - transformed_train_features['Hillshade_3pm']
)

transformed_valid_features['Hillshade_Diff'] = (
    transformed_valid_features['Hillshade_9am'] - transformed_valid_features['Hillshade_3pm']
)

transformed_test_features['Hillshade_Diff'] = (
    transformed_test_features['Hillshade_9am'] - transformed_test_features['Hillshade_3pm']
)
```

```python
# Create an interaction feature by multiplying Slope and Elevation
transformed_train_features['Slope_Elevation'] = transformed_train_features['Slope'] * transformed_train_features['Elevation']

transformed_valid_features['Slope_Elevation'] = transformed_valid_features['Slope'] * transformed_valid_features['Elevation']

transformed_test_features['Slope_Elevation'] = transformed_test_features['Slope'] * transformed_test_features['Elevation']
```

```python
# Create a feature representing the average hillshade across three time periods
transformed_train_features['Hillshade_Avg'] = (
    transformed_train_features[['Hillshade_9am', 'Hillshade_Noon', 'Hillshade_3pm']].mean(axis=1)
)

transformed_valid_features['Hillshade_Avg'] = (
    transformed_valid_features[['Hillshade_9am', 'Hillshade_Noon', 'Hillshade_3pm']].mean(axis=1)
)

transformed_test_features['Hillshade_Avg'] = (
    transformed_test_features[['Hillshade_9am', 'Hillshade_Noon', 'Hillshade_3pm']].mean(axis=1)
)
```

```python
# Create a ratio-based feature representing the relationship between roadways and hydrology distances
transformed_train_features['Roads_to_Hydrology'] = (
    transformed_train_features['Horizontal_Distance_To_Roadways'] /
    (transformed_train_features['Horizontal_Distance_To_Hydrology'] + 1)
)

transformed_valid_features['Roads_to_Hydrology'] = (
    transformed_valid_features['Horizontal_Distance_To_Roadways'] /
    (transformed_valid_features['Horizontal_Distance_To_Hydrology'] + 1)
)

transformed_test_features['Roads_to_Hydrology'] = (
    transformed_test_features['Horizontal_Distance_To_Roadways'] /
    (transformed_test_features['Horizontal_Distance_To_Hydrology'] + 1)
)
```

```python
# Create a feature representing the combined distance from hydrology and fire points
transformed_train_features['Hydro_Fire_Sum'] = (
    transformed_train_features['Horizontal_Distance_To_Hydrology'] +
    transformed_train_features['Horizontal_Distance_To_Fire_Points']
)

transformed_valid_features['Hydro_Fire_Sum'] = (
    transformed_valid_features['Horizontal_Distance_To_Hydrology'] +
    transformed_valid_features['Horizontal_Distance_To_Fire_Points']
)

transformed_test_features['Hydro_Fire_Sum'] = (
    transformed_test_features['Horizontal_Distance_To_Hydrology'] +
    transformed_test_features['Horizontal_Distance_To_Fire_Points']
)
```

```python
# Create categorical elevation range features based on elevation thresholds
transformed_train_features["elev_low"] = (transformed_train_features["Elevation"] < 2000)

transformed_train_features["elev_mid"] = (
    (transformed_train_features["Elevation"] >= 2000) &
    (transformed_train_features["Elevation"] < 3000)
)

transformed_train_features["elev_high"] = (transformed_train_features["Elevation"] >= 3000)

transformed_valid_features["elev_low"] = (transformed_valid_features["Elevation"] < 2000)

transformed_valid_features["elev_mid"] = (
    (transformed_valid_features["Elevation"] >= 2000) &
    (transformed_valid_features["Elevation"] < 3000)
)

transformed_valid_features["elev_high"] = (transformed_valid_features["Elevation"] >= 3000)

transformed_test_features["elev_low"] = (transformed_test_features["Elevation"] < 2000)

transformed_test_features["elev_mid"] = (
    (transformed_test_features["Elevation"] >= 2000) &
    (transformed_test_features["Elevation"] < 3000)
)

transformed_test_features["elev_high"] = (transformed_test_features["Elevation"] >= 3000)
```

```python
import numpy as np

# Create sine and cosine features to represent the circular nature of Aspect
transformed_train_features["north_south"] = np.cos(np.radians(transformed_train_features["Aspect"]))
transformed_train_features["east_west"] = np.sin(np.radians(transformed_train_features["Aspect"]))

transformed_valid_features["north_south"] = np.cos(np.radians(transformed_valid_features["Aspect"]))
transformed_valid_features["east_west"] = np.sin(np.radians(transformed_valid_features["Aspect"]))

transformed_test_features["north_south"] = np.cos(np.radians(transformed_test_features["Aspect"]))
transformed_test_features["east_west"] = np.sin(np.radians(transformed_test_features["Aspect"]))
```

```python
# Create a feature representing the difference between elevation and hydrology distance
transformed_train_features["elev_hydrology"] = (
    transformed_train_features["Elevation"] -
    transformed_train_features["Horizontal_Distance_To_Hydrology"]
)

transformed_valid_features["elev_hydrology"] = (
    transformed_valid_features["Elevation"] -
    transformed_valid_features["Horizontal_Distance_To_Hydrology"]
)

transformed_test_features["elev_hydrology"] = (
    transformed_test_features["Elevation"] -
    transformed_test_features["Horizontal_Distance_To_Hydrology"]
)
```

```python
# Create elevation-based features using vertical distance to hydrology
transformed_train_features['Elevation_VDH'] = (
    transformed_train_features['Elevation'] -
    transformed_train_features['Vertical_Distance_To_Hydrology']
)

transformed_train_features['Elevation_VDH_abs'] = abs(
    transformed_train_features['Vertical_Distance_To_Hydrology']
)

transformed_valid_features['Elevation_VDH'] = (
    transformed_valid_features['Elevation'] -
    transformed_valid_features['Vertical_Distance_To_Hydrology']
)

transformed_valid_features['Elevation_VDH_abs'] = abs(
    transformed_valid_features['Vertical_Distance_To_Hydrology']
)

transformed_test_features['Elevation_VDH'] = (
    transformed_test_features['Elevation'] -
    transformed_test_features['Vertical_Distance_To_Hydrology']
)

transformed_test_features['Elevation_VDH_abs'] = abs(
    transformed_test_features['Vertical_Distance_To_Hydrology']
)
```

```python
# Create a combined Euclidean distance feature for horizontal and vertical hydrology distances
transformed_train_features['Hydrology_Distance'] = np.sqrt(
    transformed_train_features['Horizontal_Distance_To_Hydrology']**2 +
    transformed_train_features['Vertical_Distance_To_Hydrology']**2
)

transformed_valid_features['Hydrology_Distance'] = np.sqrt(
    transformed_valid_features['Horizontal_Distance_To_Hydrology']**2 +
    transformed_valid_features['Vertical_Distance_To_Hydrology']**2
)

transformed_test_features['Hydrology_Distance'] = np.sqrt(
    transformed_test_features['Horizontal_Distance_To_Hydrology']**2 +
    transformed_test_features['Vertical_Distance_To_Hydrology']**2
)
```

```python
# Create aggregate distance features from roadway, fire point, and hydrology distances
transformed_train_features['Mean_Distance'] = (
    transformed_train_features['Horizontal_Distance_To_Roadways'] +
    transformed_train_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_train_features['Horizontal_Distance_To_Hydrology']
) / 3

transformed_train_features['Sum_Distance'] = (
    transformed_train_features['Horizontal_Distance_To_Roadways'] +
    transformed_train_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_train_features['Horizontal_Distance_To_Hydrology']
)

transformed_valid_features['Mean_Distance'] = (
    transformed_valid_features['Horizontal_Distance_To_Roadways'] +
    transformed_valid_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_valid_features['Horizontal_Distance_To_Hydrology']
) / 3

transformed_valid_features['Sum_Distance'] = (
    transformed_valid_features['Horizontal_Distance_To_Roadways'] +
    transformed_valid_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_valid_features['Horizontal_Distance_To_Hydrology']
)

transformed_test_features['Mean_Distance'] = (
    transformed_test_features['Horizontal_Distance_To_Roadways'] +
    transformed_test_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_test_features['Horizontal_Distance_To_Hydrology']
) / 3

transformed_test_features['Sum_Distance'] = (
    transformed_test_features['Horizontal_Distance_To_Roadways'] +
    transformed_test_features['Horizontal_Distance_To_Fire_Points'] +
    transformed_test_features['Horizontal_Distance_To_Hydrology']
)
```
