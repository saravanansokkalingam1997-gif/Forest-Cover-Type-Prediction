```python
from imblearn.over_sampling import SMOTE
import pandas as pd

# Create copies of the training features and labels for resampling
resampling_train_features = transformed_train_features.copy()
resampling_train_labels = encoded_train_labels.copy()

# Initialize normal SMOTE
smote_resampler = SMOTE(
    sampling_strategy='not majority',
    random_state=42
)

# Apply SMOTE only to the training data
resampled_train_features, resampled_train_labels = (
    smote_resampler.fit_resample(
        resampling_train_features,
        resampling_train_labels
    )
)

# Compare class distributions before and after SMOTE
print(
    "Before resampling:\n",
    pd.Series(resampling_train_labels).value_counts()
    .sort_index()
)

print(
    "\nAfter SMOTE:\n",
    pd.Series(resampled_train_labels).value_counts()
    .sort_index()
)
```
